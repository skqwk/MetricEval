### Вспомогательный блокнот

```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg

template = '../dataset-generator/experiments/{}/triplet_heatmap_simple.png'

titles = ["Исходный пример", "Логнормальное распределение длительностей запросов", "RPM = 5", "RPM = 15"]

fig, axes = plt.subplots(2, 2, figsize=(16, 16))

axes_flat = axes.flatten()

for i, ax in enumerate(axes_flat):
    img = mpimg.imread(template.format(i + 1))
    ax.imshow(img)
    ax.set_title(titles[i])
    ax.axis('off')

plt.tight_layout()
plt.show()
```
